Alle som programmerer, får feilmeldinger hele tiden, også erfarne utviklere. Det viktigste er ikke å unngå feil, men å kunne **lese feilmeldingen** og finne ut hva som er galt.

## Slik leser du en feilmelding

```
---------------------------------------------------------------------------
TypeError                                 Traceback (most recent call last)
Cell In[3], line 2
      1 alder = 16
----> 2 print("Du er " + alder + " år")

TypeError: can only concatenate str (not "int") to str
```

1. **Se på den siste linja først.** Den sier hvilken *type* feil det er (`TypeError`) og en kort forklaring.
2. **Finn linjenummeret.** Pila `---->` viser hvor Python oppdaget feilen.
3. **Les koden på den linja.** Her prøver vi å legge sammen tekst (`"Du er "`) og et tall (`alder`), og det kan ikke Python.

## Vanlige feiltyper

| Feiltype | Betyr | Typisk årsak |
|---|---|---|
| `SyntaxError` | Python forstår ikke koden | Manglende `"`, `)` eller `:` |
| `IndentationError` | Feil innrykk | Mellomrom foran en linje som ikke skal ha det |
| `NameError` | Navnet finnes ikke | Skrivefeil, eller variabelen er ikke laget ennå |
| `TypeError` | Feil datatype | Tekst + tall, eller feil antall argumenter |
| `ValueError` | Riktig type, men ugyldig verdi | `int("hei")` eller `int("3.5")` |
| `ZeroDivisionError` | Deling på null | Nevneren er 0 |

Noen feil gir **ingen** feilmelding: programmet kjører, men svaret er feil. Dette kalles **logiske feil**, og de er ofte de vanskeligste å finne.
